# Prepare the RNA oracle support

Preview the RNA prediction cache or construct its top-200,000 reward states with
full-state normalization. **The default reads only 20 entries.** Exhaustive
preparation scans all `4**14` predictions and can take substantial time.

The cache and reward-statistics file are required. The streaming parser does not
load the entire cache or unpickle arbitrary objects. See the
[implementation](analysis_utils.py) and [data instructions](../../README.md#optional-rna-oracle-preparation).

```python
from pathlib import Path
import sys
from IPython.display import display

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents]
            if (p / "code/train.py").is_file() and (p / "requirements.txt").is_file())
for folder in (ROOT / "code", ROOT / "code/notebooks"):
    if str(folder) not in sys.path:
        sys.path.insert(0, str(folder))
from analysis_utils import OraclePreparation
```

## 1. Configure the input and output

Leave values as `None` to use `GF_RNA_PREDICTION_CACHE`, `GF_RNA_ORACLE_OUTPUT`,
and `GF_PREPARE_ORACLE`. Full preparation requires `PREPARE=True` or
`GF_PREPARE_ORACLE=1`; existing oracle artifacts are never overwritten.

```python
CACHE = None    # path to L14_RNA1_allpreds.pkl
OUTPUT = None   # default outputs/rna-oracle, inside the repository
PREPARE = None  # defaults to preview; True enables the full scan

oracle = OraclePreparation(root=ROOT, cache=CACHE, output=OUTPUT, prepare=PREPARE)
display(oracle.describe())
```

## 2. Preview or prepare

Preparation verifies ordered, unique RNA keys and exactly `4**14` entries.
Rewards use the clipped score, exponent 40, and saved scale. Output metadata
records cache/statistics hashes and global reward normalization; the oracle's
captured mass is not automatically 100%.

```python
result = oracle.run()
display(result)
```
